```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


numbers = [8,6,9,1,5,16,4]

print(merge_sort(numbers))
```

```python
def mean(arr):
    sum = 0
    for i in range(len(arr)):
        sum += arr[i]
    return float(sum/len(arr))

def median(arr):
    arr.sort()
    return arr[len(arr)//2]

def mode(arr):
    freq = {}
    for i in range(len(arr)):
        if arr[i] not in freq:
            freq[arr[i]] = 1
        else:
            freq[arr[i]] += 1
    max = 0
    index = -1
    for key, value in freq.items():
        if value > max:
            max = value
            index = key
    return index

        
numbers = [8,6,9,1,5,16,4,5]
print(mean(numbers))
print(median(numbers))
print(mode(numbers))
```

```python
def counter(arr):
    freq = dict.fromkeys(arr, 0)
    for num in arr:
        freq[num] += 1
    
    index = []
    for key, value in freq.items():
        if value > 1:
            max = value
            index.append(key)
    return index
numbers = [8,6,9,1,5,16,4,5,10,10,1]
print(counter(numbers))
```
