```python
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'D'],
    'D': ['B', 'C', 'E'],
    'E': ['D', 'F'],
    'F': ['E']
}

def find_path(graph, start, goal, path=None):
    if path is None:
        path = []
    path = path + [start]
    
    if start == goal:
        return path
    if start not in graph:
        return None
        
    for neighbor in graph[start]:
        if neighbor not in path:
            new_path = find_path(graph, neighbor, goal, path)
            if new_path is not None:
                return new_path
    return None

test_cases = [('A', 'F'), ('B', 'E'), ('A', 'Z')]

for start, goal in test_cases:
    path = find_path(graph, start, goal)
    if path:
        print(f"Path from {start} to {goal}: {' -> '.join(path)}")
    else:
        print(f"No path exists from {start} to {goal}")
```

```python
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'D', 'F'],
    'D': ['B', 'C', 'E'],
    'E': ['D', 'F'],
    'F': ['C', 'E']
}

def find_all_paths(graph, start, goal, path=None):
    if path is None:
        path = []
    path = path + [start]
    
    if start == goal:
        return [path]
    if start not in graph:
        return []
        
    paths = []
    for neighbor in graph[start]:
        if neighbor not in path:
            new_paths = find_all_paths(graph, neighbor, goal, path)
            paths.extend(new_paths)
    return paths

paths = find_all_paths(graph, 'A', 'F')

for i, p in enumerate(paths, 1):
    print(f"Path {i}: {' -> '.join(p)}")
print(f"Total paths: {len(paths)}")
```

```python
# Directed weighted graph
graph = {
    'A': [('B', 2), ('C', 6)],
    'B': [('D', 5), ('E', 2)],
    'C': [('E', 2)],
    'D': [('F', 3)],
    'E': [('F', 3)],
    'F': []
}

def find_all_weighted_paths(graph, start, goal, path=None, cost=0):
    if path is None:
        path = []
    path = path + [start]
    
    if start == goal:
        return [(path, cost)]
    if start not in graph:
        return []
        
    all_paths = []
    for neighbor, weight in graph[start]:
        if neighbor not in path:
            all_paths.extend(find_all_weighted_paths(graph, neighbor, goal, path, cost + weight))
    return all_paths

paths = find_all_weighted_paths(graph, 'A', 'F')
shortest = min(paths, key=lambda x: x[1])

print("All Routes and Costs:")
for p, c in paths:
    print(f"  {' -> '.join(p)} | Cost: {c}")

print(f"\nShortest Path: {' -> '.join(shortest[0])}")
print(f"Total Cost: {shortest[1]}")
```

```python
campus_graph = {
    'Main Gate': [('Reception', 50)],
    'Reception': [('Library', 100), ('Cafeteria', 120), ('Lab', 150)],
    'Library': [('Lab', 60)],
    'Cafeteria': [('Department', 80)],
    'Department': [('Parking', 90)],
    'Lab': [('Department', 70)],
    'Parking': []
}

def get_campus_routes(graph, start, goal, path=None, dist=0):
    if path is None:
        path = []
    path = path + [start]
    if start == goal:
        return [(path, dist)]
    if start not in graph:
        return []
    
    routes = []
    for neighbor, d in graph[start]:
        if neighbor not in path:
            routes.extend(get_campus_routes(graph, neighbor, goal, path, dist + d))
    return routes

routes = get_campus_routes(campus_graph, 'Main Gate', 'Lab')

print("All Possible Routes:")
for r, d in routes:
    print(f"  {' -> '.join(r)} (Distance: {d} m)")

best_route = min(routes, key=lambda x: x[1])
print(f"\nShortest Route: {' -> '.join(best_route[0])}")
print(f"Total Distance: {best_route[1]} m")
```

```python
social_network = {
    'Ali': ['Ahmed', 'Sara'],
    'Ahmed': ['Ali', 'Hassan'],
    'Hassan': ['Ahmed', 'Fatima'],
    'Sara': ['Ali', 'Fatima'],
    'Fatima': ['Sara', 'Hassan']
}

def get_friends(graph, person):
    return graph.get(person, [])

def are_directly_connected(graph, p1, p2):
    return p2 in graph.get(p1, [])

def find_all_connection_paths(graph, start, goal, path=None):
    if path is None:
        path = []
    path = path + [start]
    if start == goal:
        return [path]
    if start not in graph:
        return []
    
    paths = []
    for neighbor in graph[start]:
        if neighbor not in path:
            paths.extend(find_all_connection_paths(graph, neighbor, goal, path))
    return paths

# 1. Friends of Ali
print("Friends of Ali:", get_friends(social_network, 'Ali'))

# 2. Check direct connections
print("Are Ali and Hassan directly connected?", are_directly_connected(social_network, 'Ali', 'Hassan'))
print("Are Ali and Ahmed directly connected?", are_directly_connected(social_network, 'Ali', 'Ahmed'))

# 3 & 4. All paths
paths = find_all_connection_paths(social_network, 'Ali', 'Hassan')
print("\nAll connection paths between Ali and Hassan:")
for p in paths:
    print(f"  {' -> '.join(p)}")

# 5. Minimum degree of separation
shortest_connection = min(paths, key=len)
print(f"\nMinimum Connection Path: {' -> '.join(shortest_connection)}")
print(f"Degree of separation (edges): {len(shortest_connection) - 1}")
```

```python
robot_rooms = {
    'A': [('B', 3), ('C', 2)],
    'B': [('D', 4)],
    'C': [('D', 1), ('E', 6)],
    'D': [('G', 5)],
    'E': [('G', 2)],
    'G': []
}

def find_robot_routes(graph, start, goal, path=None, cost=0):
    if path is None:
        path = []
    path = path + [start]
    if start == goal:
        return [(path, cost)]
    if start not in graph:
        return []
    
    routes = []
    for neighbor, c in graph[start]:
        if neighbor not in path:
            routes.extend(find_robot_routes(graph, neighbor, goal, path, cost + c))
    return routes

routes = find_robot_routes(robot_rooms, 'A', 'G')

print("All Robot Routes:")
for r, c in routes:
    print(f"  {' -> '.join(r)} | Cost: {c}")

best_route = min(routes, key=lambda x: x[1])
print(f"\nRobot's Best Route: {' -> '.join(best_route[0])}")
print(f"Minimum Movement Cost: {best_route[1]}")
```
