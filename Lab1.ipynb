```python
def sundaes():
    total = 0
    flavors = ["vanilla", "chocolate", "strawberry", "pistacchio"] 
    sauces = ["caramel", "butterscotch", "chocolate"]
    for flavor in flavors:
        for sauce in sauces:
            print(f"{flavor} ice cream sundae with {sauce} sauce.")
            total += 1
    print(f"Total number of sundaes possible: {total}")
```

```python
def triangle():
    limit = 10
    value = 1
    row = 1
    while row <= limit:
        coloumn = 1
        while coloumn <= row:
            if coloumn != row:
                print(f"{value} ",sep='', end='')
            else:
                print(value)
            value += 1
            coloumn += 1
        row += 1
triangle()
```

```python
def factorial(num):
    if num<0:
        print("Enter a positive number")
    if num==0:
        return 1
    return num*factorial(num-1)
factorial(5)
```

```python
def countpattern(lst1, lst2):
    count = 0
    for i in range(0,len(lst2)):
        start = i
        flag = True
        for j in range (len(lst1)):
            if start >= len(lst2):
                flag = False
                break
            if lst2[start] != lst1[j]:
                flag = False
            start += 1
        if flag == True:
            count += 1
    return count
    
list1 = ['a', 'b', 'a']
list2 = ['g','a','b','a','b','a','b','a']
list3 = ['a','b']
list4 = ['a','b','c','e','b','a','b','f']
print(countpattern(list1, list2))   
print(countpattern(list3, list4))   
```

```python
class roman:
    def __init__(self, num):
        self.number = str(num)
        
    def conversion(self):
        num = int(self.number)
        final = ''
        while num > 0:
            if num >= 1000:
                final += "M"
                num -= 1000
            elif num >= 900:
                final += "CM"
                num -= 900
            elif num >= 500:
                final += "D"
                num -= 500
            elif num >= 400:
                final += "CD"
                num -= 400
            elif num >= 100:
                final += "C"
                num -= 100
            elif num >= 90:
                final += "XC"
                num -= 90
            elif num >= 50:
                final += "L"
                num -= 50
            elif num >= 40:
                final += "XL"
                num -= 40
            elif num >= 10:
                final += "X"
                num -= 10
            elif num >= 9:
                final += "IX"
                num -= 9
            elif num >= 5:
                final += "V"
                num -= 5
            elif num >= 4:
                final += "IV"
                num -= 4
            elif num >= 1:
                final += "I"
                num -= 1
        self.number = final
        
numbers = roman(192)
numbers.conversion()
print(numbers.number)
```

```python
class numbers:
    def __init__(self, numbers):
        self.numbers = numbers
    
    def findpair(self, target):
        nums = self.numbers
        for i in range(len(nums)):
            if nums[i] == target:
                return [i]
            elif i+1 == range(len(nums)):
                return [-1]
            elif nums[i] < target:
                for j in range(i+1, len(nums)):
                    if nums[i] + nums[j] == target:
                        return [i,j]

target = int(input("Enter the Number to Find: "))
nums = [10,20,10,40,50,60,70]
n = numbers(nums)
print(n.findpair(target))
```

```python
class zero:
    def __init__(self, array):
        self.array = array

    def findzero(self):
        solution = []
        array = self.array
        size = len(array)
        for i in range(size):
            for j in range(i+1, size):
                for k in range(j+1, size):
                    if array[i] + array[j] + array[k] == 0:
                        solution.append([array[i], array[j], array[k]])
        return solution
                        

z = zero([-25, -10, -7, -3, 2, 4, 8, 10])
print(z.findzero())
```

```python
class sentence:
    def __init__(self, words):
        self.words = words

    def reverse(self):
        words = self.words.split()
        return [words[i] for i in range(len(words)-1, -1, -1)]

words = input("Enter a sentence: ")
s = sentence(words)
print(s.reverse())
```

```python
def multiplication(rows, cols):
    matrix1 = []
    matrix2 = []
    result = []
    
    print("Enter elements of first matrix:")
    
    for i in range(rows):
        row = []
        for j in range(cols):
            row.append(int(input()))
        matrix1.append(row)
    
    print("Enter elements of second matrix:")
    
    for i in range(rows):
        row = []
        for j in range(cols):
            row.append(int(input()))
        matrix2.append(row)
    
    for i in range(rows):
        row = []
        for j in range(cols):
            row.append(matrix1[i][j] * matrix2[i][j])
        result.append(row)
    return result

rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))

print(multiplication(rows, cols))
```

```python
def calculator(num1, num2, op='+', format="floating"):
    result = float(0)
    try:
        if op == '+':
            result = num1+num2
        elif op == '-':
            result = num1-num2
        elif op == '*':
            result = num1*num2
        elif op == '/':
            result = num1/num2
        else:
            raise ValueError("Invalid Operator Used.")
    except:
        return print("Invalid Operator Used.")
    try:   
        if format == "integer" or format == "int":
            return int(result + 0.5) if result >= 0 else int(result - 0.5)
        elif format == "floating":
            return result
        else:
            raise TypeError("Invalid Format Used")
    except:
        print("Invalid Format Used")
    
calculator(5,2.7,'/', format="inty")
```

```python
class Numbers:
    MULTIPLIER = 10

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def add(self):
        return self.x + self.y

    @classmethod
    def multiply(cls, a):
        return a * cls.MULTIPLIER

    @staticmethod
    def subtract(b, c):
        return b - c

    @property
    def value(self):
        return (self.x, self.y)

    @value.setter
    def value(self, values):
        self.x = values[0]
        self.y = values[1]

    @value.deleter
    def value(self):
        del self.x
        del self.y

n = Numbers(1,3)
print(n.add())
print(n.multiply(10))
print(Numbers.subtract(10,5))
print(n.value)
n.value = (10,5)
print(n.value)
del n.value
```
