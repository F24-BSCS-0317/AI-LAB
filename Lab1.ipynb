```python
def sundaes():
    total = 0
    flavors = ["vanilla", "chocolate", "strawberry", "pistacchio"] 
    sauces = ["caramel", "butterscotch", "chocolate"]
    for flavor in flavors:
        for sauce in sauces:
            print(f"{flavor} ice cream sundae with {sauce} sauce.")
            total += 1
    print(f"Total number of sundaes possible: {total}")
```

```python
sundaes()
```

```python
def triangle():
    limit = 10
    value = 1
    row = 1
    while row <= limit:
        coloumn = 1
        while coloumn <= row:
            if coloumn != row:
                print(f"{value} ",sep='', end='')
            else:
                print(value)
            value += 1
            coloumn += 1
        row += 1
triangle()
```

```python
def factorial(num):
    if num<0:
        print("Enter a positive number")
    if num==0:
        return 1
    return num*factorial(num-1)
factorial(5)
```

```python
def countpattern(lst1, lst2):
    count = 0
    for i in range(0,len(lst2)):
        start = i
        flag = True
        for j in range (len(lst1)):
            if start >= len(lst2):
                flag = False
                break
            if lst2[start] != lst1[j]:
                flag = False
            start += 1
        if flag == True:
            count += 1
    return count
    
list1 = ['a', 'b', 'a']
list2 = ['g','a','b','a','b','a','b','a']
list3 = ['a','b']
list4 = ['a','b','c','e','b','a','b','f']
print(countpattern(list1, list2))   
print(countpattern(list3, list4))   
```

```python
def Calculator():
    num1 = int(input())
    op = input()
    num2 = int(input())
    if op == '+':
        return num1+num2
    if op == '-':
        return num1-num2
    if op == '*':
        return num1*num2
    if op == '/':
        return num1/num2
print(Calculator())
```

```python
class roman:
    def __init__(self, num):
        self.number = str(num)
        
    def conversion(self):
        num = int(self.number)
        final = ''
        while num > 0:
            if num >= 1000:
                final += "M"
                num -= 1000
            elif num >= 900:
                final += "CM"
                num -= 900
            elif num >= 500:
                final += "D"
                num -= 500
            elif num >= 400:
                final += "CD"
                num -= 400
            elif num >= 100:
                final += "C"
                num -= 100
            elif num >= 90:
                final += "XC"
                num -= 90
            elif num >= 50:
                final += "L"
                num -= 50
            elif num >= 40:
                final += "XL"
                num -= 40
            elif num >= 10:
                final += "X"
                num -= 10
            elif num >= 9:
                final += "IX"
                num -= 9
            elif num >= 5:
                final += "V"
                num -= 5
            elif num >= 4:
                final += "IV"
                num -= 4
            elif num >= 1:
                final += "I"
                num -= 1
        self.number = final
        
numbers = roman(192)
numbers.conversion()
print(numbers.number)
```
